```python
import sympy as sp

# --- Symbols ---
rhoL, rhoR = sp.symbols('rhoL rhoR')
epsL, epsR = sp.symbols('epsL epsR')
pL, pR = sp.symbols('pL pR')

vxL, vyL, vzL = sp.symbols('vxL vyL vzL')
vxR, vyR, vzR = sp.symbols('vxR vyR vzR')

BxL, ByL, BzL = sp.symbols('BxL ByL BzL')
BxR, ByR, BzR = sp.symbols('BxR ByR BzR')

# Lorentz factors
v2L = vxL**2 + vyL**2 + vzL**2
v2R = vxR**2 + vyR**2 + vzR**2

wL = 1 / sp.sqrt(1 - v2L)
wR = 1 / sp.sqrt(1 - v2R)

# --- Magnetic field invariants ---
def compute_b2(vx, vy, vz, Bx, By, Bz, w):
    vdotB = vx*Bx + vy*By + vz*Bz
    return (Bx**2 + By**2 + Bz**2)/w**2 + vdotB**2

b2L = compute_b2(vxL, vyL, vzL, BxL, ByL, BzL, wL)
b2R = compute_b2(vxR, vyR, vzR, BxR, ByR, BzR, wR)

# --- Small b^mu ---
def compute_smallb(vx, vy, vz, Bx, By, Bz, w):
    vdotB = vx*Bx + vy*By + vz*Bz
    bt = w * vdotB
    bx = Bx/w + w * vdotB * vx
    by = By/w + w * vdotB * vy
    bz = Bz/w + w * vdotB * vz
    return bt, bx, by, bz

btL, bxL, byL, bzL = compute_smallb(vxL, vyL, vzL, BxL, ByL, BzL, wL)
btR, bxR, byR, bzR = compute_smallb(vxR, vyR, vzR, BxR, ByR, BzR, wR)

# --- Thermodynamics ---
hL, hR = sp.symbols('hL hR')   # specific enthalpy (leave symbolic)

rhb2L = rhoL*hL + b2L
rhb2R = rhoR*hR + b2R

Pb2L = pL + b2L/2
Pb2R = pR + b2R/2

# --- Conserved quantities ---
DL = rhoL * wL
DR = rhoR * wR

# Momentum (S_i)
SxL = rhb2L * wL**2 * vxL - btL * bxL
SyL = rhb2L * wL**2 * vyL - btL * byL
SzL = rhb2L * wL**2 * vzL - btL * bzL

SxR = rhb2R * wR**2 * vxR - btR * bxR
SyR = rhb2R * wR**2 * vyR - btR * byR
SzR = rhb2R * wR**2 * vzR - btR * bzR

# Energy (tau)
tauL = rhb2L * wL**2 - Pb2L - btL**2 - DL
tauR = rhb2R * wR**2 - Pb2R - btR**2 - DR

# --- Fluxes in x-direction (idir = 0) ---
vdirL = vxL
vdirR = vxR
bdirL = bxL
bdirR = bxR

# D flux
fDL = DL * vdirL
fDR = DR * vdirR

# tau flux
ftauL = rhb2L * wL**2 * vdirL - btL * bdirL - DL * vdirL
ftauR = rhb2R * wR**2 * vdirR - btR * bdirR - DR * vdirR

# Momentum fluxes
fSxL = rhb2L * wL**2 * vxL * vdirL + Pb2L - bdirL * bxL
fSyL = rhb2L * wL**2 * vyL * vdirL - bdirL * byL
fSzL = rhb2L * wL**2 * vzL * vdirL - bdirL * bzL

fSxR = rhb2R * wR**2 * vxR * vdirR + Pb2R - bdirR * bxR
fSyR = rhb2R * wR**2 * vyR * vdirR - bdirR * byR
fSzR = rhb2R * wR**2 * vzR * vdirR - bdirR * bzR

# Magnetic field fluxes
fBxL = 0
fByL = vxL*ByL - vyL*BxL
fBzL = vxL*BzL - vzL*BxL

fBxR = 0
fByR = vxR*ByR - vyR*BxR
fBzR = vxR*BzR - vzR*BxR
```

```python
def uHLL(uL,uR,FL,FR,cmin,cmax):
    return (cmax*uR - cmin*uL + FL - FR)/(cmax-cmin)
def fHLL(uL,uR,FL,FR,cmin,cmax):
    return (cmax*FL - cmin*FR + cmax*cmin*(uR-uL))/(cmax-cmin)
```

```python
tauHLL = uHLL(tauL, tauR, ftauL, ftauR, -1, 1)
FtauHLL = fHLL(tauL, tauR, ftauL, ftauR, -1, 1)
DHLL = uHLL(DL, DR, fDL, fDR, -1, 1)
FDHLL = fHLL(DL, DR, fDL, fDR, -1, 1)

SxHLL = uHLL(SxL, SxR, fSxL, fSxR, -1, 1)
FSxHLL = fHLL(SxL, SxR, fSxL, fSxR, -1, 1)

ByHLL = uHLL(ByL, ByR, fByL, fByR, -1, 1)
FByHLL = fHLL(ByL, ByR, fByL, fByR, -1, 1)
BzHLL = uHLL(BzL, BzR, fBzL, fBzR, -1, 1)
FBzHLL = fHLL(BzL, BzR, fBzL, fBzR, -1, 1)

# Rootfinding vor v_star
a = FtauHLL + FDHLL - ByHLL * FByHLL - BzHLL * FBzHLL
b = -(FSxHLL + DHLL + tauHLL) + ByHLL * ByHLL + BzHLL * BzHLL + FByHLL * FByHLL + FBzHLL * FBzHLL
c = SxHLL - ByHLL * FByHLL - BzHLL * FBzHLL
v_star = (-b - sp.sqrt(b**2 - 4*a*c))/(2*a)
v_star.simplify()
```
